```python
category_revenue = df.groupby("Category")["Revenue"].sum()

x = category_revenue.index
y = category_revenue.values

plt.bar(x, y)
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.title("Revenue by Category")
plt.show()
```

```python
monthly_revenue = df.groupby("Month")["Revenue"].sum()
x=monthly_revenue.index
y=monthly_revenue.values
plt.plot(x, y)
plt.xlabel("Months")
plt.ylabel("Revenue")
plt.title("Monthly Revenue Trend")
plt.show()
```

```python
region_revenue = df.groupby("Region")["Revenue"].sum()

x = region_revenue.index
y = region_revenue.values

plt.bar(x, y)
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.title("Region-wise Revenue")
plt.show()
```

```python
salesperson_profit = df.groupby("Salesperson")["Profit"].sum()

x = salesperson_profit.index
y = salesperson_profit.values

plt.bar(x, y)
plt.xlabel("Salesperson")
plt.ylabel("Profit")
plt.title("Salesperson-wise Profit")
plt.show()
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum().sort_values(ascending=False)

x = product_revenue.index
y = product_revenue.values

plt.bar(x, y)
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.title("Product-wise Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
total = df[["Revenue", "Profit"]].sum()

x = total.index
y = total.values

plt.bar(x, y)
plt.xlabel("Metrics")
plt.ylabel("Amount (₹)")
plt.title("Total Revenue vs Total Profit")
plt.show()
```

```python
payment_count = df["Payment_Method"].value_counts()

plt.pie(payment_count, labels=payment_count.index, autopct="%1.1f%%")
plt.title("Payment Method Distribution")
plt.show()
```
